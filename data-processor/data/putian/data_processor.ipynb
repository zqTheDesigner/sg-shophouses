```python
import json

# Input and output files
INPUT_FILE = "./raw/Ancient_Shoreline_Inverted.json"
OUTPUT_FILE = "Ancient_Shoreline_Inverted.json"

# Load GeoJSON
with open(INPUT_FILE, "r", encoding="utf-8") as f:
    geojson = json.load(f)

converted = []

for feature in geojson["features"]:
    props = feature["properties"].copy()

    # Remove unwanted fields
    props.pop("Id", None)

    # Rename geometry field
    props["GEOMETRY"] = feature["geometry"]

    converted.append(props)

# Save
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    json.dump(converted, f, indent=4)

print(f"Converted {len(converted)} polygons.")
```
